```
import cv2
from matplotlib import pyplot as plt
```

```
img = cv2.imread('image.jpeg')

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```
roi = img[200:400, 250:450]
```

```
img_copy = img.copy()

img_copy[0:200, 0:200] = roi
```

```
cv2.rectangle(img, (250,150), (450,350), (255,0,0), 4)
```

```
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

result_rgb = cv2.cvtColor(img_copy, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("Original Image with ROI")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(result_rgb)
plt.title("ROI Copied and Pasted")
plt.axis("off")

plt.show()
```
